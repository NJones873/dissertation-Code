```python
import numpy as np
import pandas as pd
import xgboost as xgb
import pickle
from tkinter import *
import csv




global names
global arr
global dat1
global post
post=[-1,0]
names=[]
dat1=[]
arr=[]


root = Tk()


root.title("Window")

root.geometry('350x200')


lbl = Label(root, text = "Start ?")
lbl.grid()


txt = Entry(root, width=10)
txt.grid(column =1, row =0)



def setup():
    global names
    global arr
    global dat1
    with open("model_for_heal.csv", 'r') as f:
        reader = csv.reader(f)
        data = list(reader) 
    data_array = np.array(data)
    names=data_array[0].tolist()
    
    
    dat1 = pd.DataFrame({'Blank': [float('nan')]})
    with open("heal_distance.csv", 'r') as f:
        reader = csv.reader(f)
        data = list(reader) 
    data_array = np.array(data)
    arr=data_array[1].tolist()



def clicked():
    global names
    global arr
    global dat1
    global post
    i=post[0]
    j=post[1]
    if i!=23 and i!=-1:
        try:
            values=float(txt.get())
        except:
            values=float('nan')
        dat2 = pd.DataFrame({names[i]: [values]})
        dat1 = pd.concat([dat1, dat2], axis=1)
        post[0]=post[0]+1
        if post[0]!=23:
            res=names[post[0]]+ ": "
            lbl.configure(text = res)
            txt.delete(0, END)
        else:
            txt.delete(0, END)
            VoteClass_1(dat1)
    elif i==-1:
        i=1
        res=names[i]+ ": "
        lbl.configure(text = res)
        post[0]=1
        
def VoteClass_1(treatment):


    X_test=treatment.iloc[:,1:]
    y_test=treatment.iloc[:,0]

    
    file_name = "xgb_class1_3.pkl"
    clf = pickle.load(open(file_name, "rb"))
    
    y_pred = clf.predict(X_test, iteration_range=(0, 12))
    for i in range (0,len(y_pred)):
        if y_pred[(i)]==0:
            var="Heal"
        else:
            var="Not"
    res=var
    lbl.configure(text = res)

#--------------------------------------------------------------------------------------

btn = Button(root, text = "Click me" ,fg = "red", command=clicked)

btn.grid(column=2, row=0)
btn.place(relx=0.5, rely=0.5, anchor=CENTER)

setup()
root.mainloop()
```
