```python
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
from sklearn import tree
import matplotlib.pyplot as plt
import re
import matplotlib
import numpy as np
import pandas as pd
import xgboost as xgb
from xgboost import plot_tree
import graphviz
from xgboost import XGBRegressor
from xgboost import XGBClassifier
import matplotlib.pyplot as plt
import random
from tensorflow.keras.utils import to_categorical
from sklearn.ensemble import VotingClassifier
from sklearn.ensemble import StackingClassifier
from sklearn.ensemble import VotingClassifier
from sklearn.preprocessing import StandardScaler
from mlxtend.plotting import plot_decision_regions
from sklearn.tree import DecisionTreeClassifier
import pickle


Treatment=pd.read_csv("../dissertation/Wound area V1_1.csv")
df_x=Treatment.iloc[:,1:]
df_y=Treatment.iloc[:,0]
R_state=random.randrange(1,10)
print("Random_state: ",R_state)
#R_state=4
X_train, X_test, y_train, y_test = train_test_split(df_x, df_y, test_size=0.15, random_state=R_state,stratify=df_y)

y_train_clf = to_categorical(y_train)
y_test_clf = to_categorical(y_test)

fullinfo=True
fulltree=False
feature_names=["IGTN","VLU","Trauma/blister","DFU","Neuropathic","Burn","Gout","Acute(0)/Chronic","WoundLength","WoundWidth","WoundArea","V1 - Image J area in mm2"]



def class1():
    clf=XGBClassifier(tree_method="hist",
                colsample_bynode=0.3,
                    colsample_bylevel=0.2,
                  colsample_bytree=0.8,
                    subsample=0.6,
                    min_child_weight=0.2,
                    objective="reg:pseudohubererror",
                    max_bin=2056,
                     eta=0.2,
                  reg_alpha=0.1,
                  num_parallel_tree=15,
                  reg_lambda=0.07,
                  n_estimators=670,
                  max_depth=None

    )
    if fullinfo==True:
        clf.fit(X_train, y_train_clf)
        y_pred = clf.predict(X_test)
        y_pred = (y_pred > 0.5) 
        print(classification_report(y_test_clf,y_pred,output_dict=False,target_names=["healing","Not-healing"]))
        
        if fulltree==True:
            importances = clf.feature_importances_
            forest_importances = pd.Series(importances, index=feature_names)
            fig, ax = plt.subplots()
            forest_importances.plot.bar( ax=ax)
            ax.set_title("Feature importances using MDI")
            ax.set_ylabel("Mean decrease in impurity")
            fig.tight_layout()
            print("---------------------------------------------------------------------")
            clf.fit(X_train, y_train)
            plt.figure(figsize=(400, 120))
            plot_tree(clf)
        accuracy = accuracy_score(y_test_clf, y_pred)
        print("Accuracy:", (accuracy*100),"%")
        print("------------------------------------------Classifier 1-------------------------------")
    return clf

def class2():
    clf=XGBClassifier(tree_method="hist",
                  eta=0.2,
                  subsample=0.5,
                 
                  # objective="reg:pseudohubererror",
                   max_bin=2056,
                    reg_alpha=0.5,
                 reg_lambda=0.4,
    )
    if fullinfo==True:
        clf.fit(X_train, y_train_clf)
        y_pred = clf.predict(X_test)
        y_pred = (y_pred > 0.5) 
        print(classification_report(y_test_clf,y_pred,output_dict=False,target_names=["healing","Not-healing"]))
    
        if fulltree==True:
            importances = clf.feature_importances_
            forest_importances = pd.Series(importances, index=feature_names)
            fig, ax = plt.subplots()
            forest_importances.plot.bar( ax=ax)
            ax.set_title("Feature importances using MDI")
            ax.set_ylabel("Mean decrease in impurity")
            fig.tight_layout()
            print("---------------------------------------------------------------------")
            clf.fit(X_train, y_train)
            plt.figure(figsize=(400, 120))
            plot_tree(clf)
        accuracy = accuracy_score(y_test_clf, y_pred)
        print("Accuracy:", (accuracy*100),"%")
        print("------------------------------------------Classifier 2-------------------------------")
    return clf

def class3():
    clf=XGBClassifier(tree_method="exact",
                  eta=0.2,
                  subsample=0.5,
                 
                   objective="reg:pseudohubererror",
                   max_bin=2056,
                    reg_alpha=0.5,
                 reg_lambda=0.4,
    )
    if fullinfo==True:
        clf.fit(X_train, y_train_clf)
        y_pred = clf.predict(X_test)
        y_pred = (y_pred > 0.5) 
        print(classification_report(y_test_clf,y_pred,output_dict=False,target_names=["healing","Not-healing"]))
    
        if fulltree==True:
            importances = clf.feature_importances_
            forest_importances = pd.Series(importances, index=feature_names)
            fig, ax = plt.subplots()
            forest_importances.plot.bar( ax=ax)
            ax.set_title("Feature importances using MDI")
            ax.set_ylabel("Mean decrease in impurity")
            fig.tight_layout()
            clf.fit(X_train, y_train)
            plt.figure(figsize=(400, 120))
            plot_tree(clf)
        accuracy = accuracy_score(y_test_clf, y_pred)
        print("Accuracy:", (accuracy*100),"%")
        print("------------------------------------------Classifier 3-------------------------------")
    return clf
    

    
def VoteClass():
    clf1=class1()
    clf2=class2()
    clf3=class3()
    sclf = VotingClassifier(estimators=[('clf1', clf1), ('clf2', clf2), ('clf3', clf3)],voting='soft',flatten_transform=False)
    
    sclf.fit(X_train, y_train)
    
    y_pred = sclf.predict(X_test)
    print(classification_report(np.asarray(y_test_clf).argmax(axis=1),y_pred,output_dict=False,target_names=["healing","Not-healing"]))
    
    accuracy = accuracy_score(np.asarray(y_test_clf).argmax(axis=1), y_pred)
    print("Accuracy:", (accuracy*100),"%")

    y_pred = sclf.predict_proba(X_test)
    print(y_pred)
    
    
    
    
VoteClass()
```
