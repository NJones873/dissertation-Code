```python
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
from sklearn import tree
import matplotlib.pyplot as plt
import re
import matplotlib
import numpy as np
import pandas as pd
import xgboost as xgb
from xgboost import plot_tree
import graphviz
from xgboost import XGBRegressor
from xgboost import XGBClassifier
import matplotlib.pyplot as plt
import random
from tensorflow.keras.utils import to_categorical
from sklearn.ensemble import VotingClassifier
from sklearn.ensemble import StackingClassifier
from sklearn.ensemble import VotingClassifier
from sklearn.preprocessing import StandardScaler
from mlxtend.plotting import plot_decision_regions
import pickle





Treatment=pd.read_csv("../dissertation/knitted dresing-prepared_V5.csv")
test=pd.read_csv("../dissertation/testss.csv")

testdf_x=test.iloc[:,1:]
testdf_y=test.iloc[:,0]


df_x=Treatment.iloc[:,1:]
df_y=Treatment.iloc[:,0]
R_state=random.randrange(1,100)
print("Random_state: ",R_state)
#R_state=23


X_train, X_test, y_train, y_test = train_test_split(df_x, df_y, test_size=0.15, random_state=R_state,stratify=df_y)
#X_train, X_test, y_train, y_test = train_test_split(df_x, df_y, test_size=0.13,stratify=df_y)

#print(X_train)
y_trainclf = to_categorical(y_train)
y_testclf = to_categorical(y_test)




clfs=XGBClassifier(tree_method="hist",
                colsample_bynode=0.3,
                    colsample_bylevel=0.2,
                  colsample_bytree=0.8,
                    subsample=0.6,
                    min_child_weight=0.2,
                    objective="reg:pseudohubererror",
                    max_bin=2056,
                     eta=0.2,
                  reg_alpha=0.1,
                  num_parallel_tree=15,
                  reg_lambda=0.07,
                  n_estimators=670,
                  max_depth=None
)

clf=XGBClassifier(tree_method="exact",
                  eta=0.2,
                  subsample=0.5,
                 
                   objective="reg:pseudohubererror",
                   max_bin=2056,
                    reg_alpha=0.5,
                 reg_lambda=0.4,
                   )

clf2=XGBClassifier(tree_method="hist",
                  eta=0.2,
                  subsample=0.5,
                 
                   objective="reg:pseudohubererror",
                   max_bin=2056,
                    reg_alpha=0.5,
                 reg_lambda=0.4,
                   )

clf1 = clf
clf2 = clf2
clf3 = clfs


clf.fit(X_train, y_trainclf)

y_pred = clf.predict(X_test)
y_pred = (y_pred > 0.5) 


print(y_pred)
print(y_testclf)

print(classification_report(y_testclf,y_pred,output_dict=False,target_names=["healing","Not-healing"]))


sclf = VotingClassifier(estimators=[('clf1', clf1), ('clf2', clf2), ('clf3', clf3)],voting='soft')
sclf.fit(X_train, y_train)
y_pred = sclf.predict(df_x)
y_pred = (y_pred > 0.5) 
print("Tree>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>")
y_test = (y_test > 0.5) 
print(classification_report(df_y,y_pred,output_dict=False,target_names=["healing","Not-healing"]))
accuracy = accuracy_score(df_y, y_pred)
print("Accuracy:", (accuracy*100),"%")
y_pred = sclf.predict_proba(df_x)
print(y_pred)



testy_pred = sclf.predict(testdf_x)
testy_pred = (testy_pred > 0.5) 
print(testy_pred)




def fullclass():

    clf.fit(X_train, y_train)
    plt.figure(figsize=(400, 120))
    plot_tree(clf)
    y_pred = clf.predict(X_test)
    y_pred = (y_pred > 0.5) 
    print(classification_report(y_test,y_pred,output_dict=False,target_names=["healing","Not-healing"]))

    clf2.fit(X_train, y_train)
    plt.figure(figsize=(400, 120))
    plot_tree(clf2)
    y_pred = clf2.predict(X_test)
    y_pred = (y_pred > 0.5) 
    print(classification_report(y_test,y_pred,output_dict=False,target_names=["healing","Not-healing"]))

    clf3.fit(X_train, y_train)
    plt.figure(figsize=(400, 120))
    plot_tree(clf3)
    y_pred = clf3.predict(X_test)
    y_pred = (y_pred > 0.5) 
    print(classification_report(y_test,y_pred,output_dict=False,target_names=["healing","Not-healing"]))
    



#fullclass()

```
